# Interactive Cobb-Douglas MPT Plot
This notebook provides an interactive plot of the Marginal Product of Land.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# Parameters
alpha = 0.3
T_values = np.linspace(10, 300, 500)

def mpt(T, L, alpha):
    return alpha * (L / T)**(1 - alpha)

def plot_mpt(L, T_bar):
    plt.figure(figsize=(10, 6))
    
    # MPT Curve
    y_values = mpt(T_values, L, alpha)
    plt.plot(T_values, y_values, label=f'MPT(T, L={L:.0f})', color='blue', linewidth=2)
    
    # Equilibrium rental rate R
    R = mpt(T_bar, L, alpha)
    
    # Vertical Land Supply
    plt.axvline(x=T_bar, color='red', linestyle='--', linewidth=2, label=f'Land Supply $\\bar{{T}}$={T_bar:.0f}')
    
    # Horizontal Indicator for Rental Rate
    plt.hlines(y=R, xmin=0, xmax=T_bar, color='green', linestyle=':', linewidth=2, label=f'Rental Rate R={R:.2f}')
    
    # Equilibrium Point
    plt.scatter([T_bar], [R], color='black', zorder=5, s=80, label='Equilibrium')
    
    # Formatting
    plt.xlim(0, 300)
    plt.ylim(0, max(1.5, R * 1.5))
    plt.title('Marginal Product of Land and Equilibrium Rental Rate')
    plt.xlabel('Land (T)')
    plt.ylabel('Rental Rate / MPT')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

# Create interactive plot with sliders
interact(plot_mpt, 
         L=FloatSlider(min=50, max=200, step=10, value=100, description='Labor (L)'), 
         T_bar=FloatSlider(min=50, max=250, step=10, value=100, description='Land Supply (T̄)'))
```
